```python
from PIL import Image
```

```python
img = Image.open("cat.png")
img.show()
```

```python
print("이미지 크기 :", img.size)
print("이미지 형식 :", img.format)
print("이미지 모드 :", img.mode)
```

```python
resize_img = img.resize((300, 300))

resize_img.show()
```

```python
print("변경 전 :", img.size)

print("변경 후 :", resize_img.size)
```

```python
crop_img = img.crop((100, 100, 500, 500))

crop_img.show()
```

```python
print("원본 크기 :", img.size)
print("자른 이미지 크기 :", crop_img.size)
```

```python
crop_img.save("crop_cat.png")
```

```python
rotate_img = img.rotate(-90)

rotate_img.show()
```

```python
rotate_img = img.rotate(45)
rotate_img.show()
```

```python
rotate_img = img.rotate(45, expand=True)
rotate_img.show()
```

```python
from PIL import Image

img = Image.open("dog.png")

img = img.resize((500, 500))

img = img.rotate(45, expand=True)

img = img.convert("L")

img.save("result.png")

img.show()

# 결과 확인
```

```python
print("=== 이미지 처리 프로그램 ===")
print("1. 이미지 크기 변경")
print("2. 이미지 자르기")
print("3. 이미지 회전")
print("4. 흑백 이미지로 변환")

menu = input("원하는 기능을 선택하세요: ")
```

```python
img = Image.open("dog.png")

if menu == "1":
    img = img.resize((300, 300))

elif menu == "2":
    img = img.crop((100, 100, 500, 500))

elif menu == "3":
    img = img.rotate(90, expand=True)

elif menu == "4":
    img = img.convert("L")
    
img.show()
img.save("result.png")

print("이미지 처리가 완료되었습니다.")
```
